```python
import RNA

# ================= 1. 输入待检测引物序列 =================
# 课题确定的 RPA Forward Primer (5' -> 3')
fwd_primer = "ACGACATCGATTTATAGCACCATCTGAAATCGGT"
rev_primer = "GTCGTATCCAGTGCAGGGTCCGAG"

# 判定阈值 (kcal/mol)
DG_THRESHOLD = -2.0


# ================= 2. 定义引物自身发卡自由能计算函数 =================
def check_primer_hairpin(primer_seq: str, primer_name: str = "Primer"):
    """使用 ViennaRNA (RNAfold) 评估引物自身发卡折叠自由能 MFE

    参数:
    ----
    primer_seq : str
        引物 DNA 序列 (5' -> 3')
    primer_name : str
        引物名称
    """
    # 转换为大写并处理空白
    seq = primer_seq.upper().strip()

    # 创建 fold_compound 对象 (ViennaRNA 支持计算单链核酸折叠)
    fc = RNA.fold_compound(seq)
    structure, mfe = fc.mfe()

    # 判断是否满足 ΔG > -2.0 kcal/mol 要求
    is_passed = mfe > DG_THRESHOLD

    print(f"=== {primer_name} 自身发卡与二级结构分析 ===")
    print(f"序列 (5'->3') : {seq}")
    print(f"序列长度      : {len(seq)} nt")
    print(f"预测二级结构  : {structure}")
    print(f"自身折叠 MFE  : {mfe:.2f} kcal/mol")
    print(f"阈值要求      : ΔG > {DG_THRESHOLD:.1f} kcal/mol")
    print(
        f"判定结果      : {'【合格 PASS】(无恶性自身发卡)' if is_passed else '【警告 FAIL】(自身发卡过强，可能阻碍退火扩增)'}"
    )
    print("-" * 50)

    return {"name": primer_name, "structure": structure, "mfe": mfe, "passed": is_passed}


# ================= 3. 运行检测 =================
res_fwd = check_primer_hairpin(fwd_primer, "RPA Forward Primer")
res_rev = check_primer_hairpin(rev_primer, "RPA Reverse Primer")
```

```python
import RNA

# 优化后的候选引物
opt_fwd_primer = "ATACTCATCATTTATAGCACCATCTGAAATCGGT"
opt_rev_primer = "GTCGTATCCAGTGCAGGGTC"

DG_THRESHOLD = -2.0


def eval_primer(seq: str, name: str):
    fc = RNA.fold_compound(seq.upper().strip())
    struct, mfe = fc.mfe()
    passed = mfe > DG_THRESHOLD

    print(f"[{name}]")
    print(f"  序列 (5'->3') : {seq}")
    print(f"  二级结构      : {struct}")
    print(f"  折叠 MFE      : {mfe:.2f} kcal/mol")
    print(
        f"  判定结果      : {'【合格 PASS】' if passed else '【未通过 FAIL】'}\n"
    )
    return passed


eval_primer(opt_fwd_primer, "优化后 RPA Forward Primer")
eval_primer(opt_rev_primer, "优化后 RPA Reverse Primer")
```

```python
import RNA

fwd_primer = "ATACTCATCATTTATAGCACCATCTGAAATCGGT"

candidates = {
    "Rev_1 (19nt, 3'滑动)": "AGTGCAGGGTCCGAGGTAT",
    "Rev_2 (18nt, 5'骨架)": "GTCGTATCCAGTGCAGGG",
    "Rev_3 (20nt, 接头区)": "TGCAGGGTCCGAGGTATTCG",
}

print("=== RPA Reverse Primer 多候选全面热力学评估 ===\n")

for name, rev_seq in candidates.items():
    # 1. 自身折叠 MFE
    fc = RNA.fold_compound(rev_seq)
    struct, self_mfe = fc.mfe()
    self_pass = self_mfe > -2.0

    # 2. 与 Fwd Primer 的双链二聚体结合能
    duplex = RNA.duplexfold(fwd_primer, rev_seq)
    dimer_energy = duplex.energy
    dimer_pass = dimer_energy > -6.0

    # 3. 检查双方 3' 末端 5 nt 是否互补 (避免引发非特异延伸/引物二聚体假阳性)
    fwd_3p = fwd_primer[-5:]
    rev_3p = rev_seq[-5:]
    duplex_3p = RNA.duplexfold(fwd_3p, rev_3p)
    end_pass = duplex_3p.energy > -1.5

    print(f"[{name}]")
    print(f"  序列 (5'->3')   : {rev_seq}")
    print(
        f"  自身结构 / MFE  : {struct}  ({self_mfe:.2f} kcal/mol) -> {'PASS' if self_pass else 'FAIL'}"
    )
    print(
        f"  与 Fwd 二聚体能 : {dimer_energy:.2f} kcal/mol -> {'PASS' if dimer_pass else 'WARN'}"
    )
    print(
        f"  3' 端末端配对能: {duplex_3p.energy:.2f} kcal/mol -> {'安全 (无 3 端锁死延伸)' if end_pass else '有 3 端二聚体风险'}"
    )
    print("-" * 55)
```
