2. 高阶生物物理特征提取def

```python
def extract_refined_features(spacer, target, mismatch_pos):
    # 1. 杂交结合自由能 Delta G duplex
    duplex = RNA.duplexfold(spacer, target)
    dg_duplex = duplex.energy

    # 2. 完全匹配对照能与扰动差值 Delta Delta G
    dg_perfect = RNA.duplexfold(spacer, spacer.replace("U", "T")).energy
    ddg = dg_duplex - dg_perfect

    # 3. 错配碱基类型表征
    pos = int(mismatch_pos)
    sp_base = spacer[pos] if pos < len(spacer) else "N"
    tg_base = target[pos] if pos < len(target) else "N"

    # 是否属于 G-T/U-G wobble 弱配对
    is_wobble = 1 if ((sp_base, tg_base) in [("G", "T"), ("U", "G")]) else 0
    
    # 是否属于强排斥同型嘌呤错配 (A-A, G-G, G-A)
    is_purine_mismatch = 1 if (sp_base in ["A", "G"] and tg_base in ["A", "G"]) else 0

    # 4. 基于 Cas12a 构象门控的位置连续高斯加权评分
    pos_weight = np.exp(-((pos - 6) ** 2) / (2 * (3.5**2)))

    return [dg_duplex, ddg, mismatch_pos, pos_weight, is_wobble, is_purine_mismatch]
```

3. 构建特征矩阵 X 与标签 y 

```python
X = np.array([
    extract_refined_features(
        row["spacer_seq"], row["target_seq"], row["mismatch_pos"]
    )
    for _, row in df.iterrows()
])
y = df["relative_activity"].values

# ================= 4. 随机森林建模 + LOOCV 评估 =================
loo = LeaveOneOut()
y_true, y_pred = [], []

rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    max_depth=4,
    min_samples_split=3,
    max_features=0.8,
)

for train_index, test_index in loo.split(X):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]

    rf.fit(X_train, y_train)
    pred = rf.predict(X_test)

    y_true.append(y_test[0])
    y_pred.append(pred[0])

# 计算评估指标
rho, p_val = spearmanr(y_true, y_pred)
mse = mean_squared_error(y_true, y_pred)

print("=== 优化后随机森林 LOOCV 评估结果 ===")
print(f"样本总数 (N): {len(df)}")
print(f"Spearman 秩相关系数 (r_s): {rho:.3f} (p-value: {p_val:.4e})")
print(f"均方误差 (MSE): {mse:.4f}")
```

```python
def extract_refined_features(spacer, target, mismatch_pos):
    pos = int(mismatch_pos)
    
    # 1. 杂交自由能
    duplex = RNA.duplexfold(spacer, target)
    dg_duplex = duplex.energy

    # 2. 修正ddg：错配target vs完美匹配target
    perfect_target = spacer.replace("T", "U")  # spacer的完美互补
    dg_perfect = RNA.duplexfold(spacer, perfect_target).energy
    ddg = dg_duplex - dg_perfect

    # 3. 错配碱基
    sp_base = spacer[pos] if pos < len(spacer) else "N"
    tg_base = target[pos] if pos < len(target) else "N"
    
    is_wobble = 1 if (sp_base, tg_base) in [("G", "T"), ("U", "G"), ("G", "U")] else 0
    is_purine_mismatch = 1 if (sp_base in "AG" and tg_base in "AG") else 0
    is_pyrimidine_mismatch = 1 if (sp_base in "CU" and tg_base in "CU") else 0

    # 4. 位置高斯权重（seed region ~1-8位对Cas12a最关键）
    pos_weight = np.exp(-((pos - 4) ** 2) / (2 * (3.0 ** 2)))
    
    # 5. 是否在seed region（位置0-7，从PAM端数）
    in_seed =1 if pos <= 7 else 0

    # 6. GC含量
    gc_spacer = (spacer.count("G") + spacer.count("C")) / len(spacer)
    
    # 7. 错配位点局部GC（±2位）
    local_seq = spacer[max(0, pos-2): pos+3]
    local_gc = (local_seq.count("G") + local_seq.count("C")) / len(local_seq) if local_seq else 0.5

    # 8. 转换/颠换类型（transition往往影响更小）
    purines = "AG"
    is_transition = 1 if (sp_base in purines) == (tg_base in purines) else 0

    return [dg_duplex, ddg, pos, pos_weight, in_seed,
            is_wobble, is_purine_mismatch, is_pyrimidine_mismatch,    gc_spacer, local_gc, is_transition]
```

```python
from sklearn.svm import SVR
from sklearn.linear_model import Ridge
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import LeaveOneOut
from scipy.stats import spearmanr
import numpy as np

# 关键：用新函数重新生成特征矩阵
X = np.array([
    extract_refined_features(
        row["spacer_seq"], row["target_seq"], row["mismatch_pos"]
    )
    for _, row in df.iterrows()
])
y = df["relative_activity"].values

print(f"新特征维度: {X.shape}")  # 应该显示 (50, 10) 而不是 (50, 6)

models = {
    "SVR":   make_pipeline(StandardScaler(), SVR(kernel="rbf", C=10, epsilon=0.05)),
    "Ridge": make_pipeline(StandardScaler(), Ridge(alpha=1.0)),
    "GBDT":  GradientBoostingRegressor(n_estimators=100, max_depth=2, learning_rate=0.05, random_state=42),
}

loo = LeaveOneOut()

for name, model in models.items():
    y_true, y_pred = [], []
    for train_idx, test_idx in loo.split(X):
        model.fit(X[train_idx], y[train_idx])
        y_pred.append(model.predict(X[test_idx])[0])
        y_true.append(y[test_idx][0])
    rho, p = spearmanr(y_true, y_pred)
    print(f"{name}: r_s = {rho:.3f} (p={p:.4e})")
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import make_scorer
from scipy.stats import spearmanr

# 自定义 Spearman 评分函数
def spearman_score(y_true, y_pred):
    rho, _ = spearmanr(y_true, y_pred)
    return rho

param_grid = {
    "svr__C": [0.1, 1, 5, 10, 50],
    "svr__epsilon": [0.01, 0.05, 0.1],
    "svr__gamma": ["scale", "auto", 0.1],
}

pipe = make_pipeline(StandardScaler(), SVR(kernel="rbf"))
gs = GridSearchCV(
    pipe, param_grid, 
    cv=5,  # 改用 5 折，LOO 在这里太慢且不稳定
    scoring=make_scorer(spearman_score),
    n_jobs=-1
)
gs.fit(X, y)

print("最佳参数:", gs.best_params_)
print("最佳 Spearman:", gs.best_score_)

# 用最佳参数再跑完整 LOO 验证
best_model = gs.best_estimator_
from sklearn.model_selection import cross_val_predict, LeaveOneOut
y_pred = cross_val_predict(best_model, X, y, cv=LeaveOneOut())
final_rho, final_p = spearmanr(y, y_pred)
print(f"LOO 最终: r_s = {final_rho:.3f}, p = {final_p:.4e}")
```

```python
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVR
from scipy.stats import spearmanr

# 只用 top 4 特征（去掉 local_gc 和 is_wobble）
top4_idx = [2, 1, 6, 10]  # mismatch_pos, ddg, is_purine_mismatch, is_transition
top4_names = ["mismatch_pos", "ddg", "is_purine_mismatch", "is_transition"]
X_top4 = X[:, top4_idx]

print(f"Top 4 特征: {top4_names}\n")

# 1. SVR (rbf) - 之前最好的
svr = make_pipeline(StandardScaler(), SVR(kernel="rbf", C=1, epsilon=0.1, gamma="scale"))
y_pred_svr = cross_val_predict(svr, X_top4, y, cv=LeaveOneOut())
rho_svr, p_svr = spearmanr(y, y_pred_svr)
print(f"SVR (4特征): r_s = {rho_svr:.3f}, p = {p_svr:.4e}")

# 2. Gradient Boosting - 通常在小样本非线性数据上表现好
gb = GradientBoostingRegressor(
    n_estimators=100, 
    max_depth=2,  # 浅树防过拟合
    learning_rate=0.05,
    subsample=0.8,
    random_state=42
)
y_pred_gb = cross_val_predict(gb, X_top4, y, cv=LeaveOneOut())
rho_gb, p_gb = spearmanr(y, y_pred_gb)
print(f"GradientBoosting (4特征): r_s = {rho_gb:.3f}, p = {p_gb:.4e}")

# 3. 非线性集成：SVR + GB 简单平均
y_pred_ensemble = (y_pred_svr + y_pred_gb) / 2
rho_ens, p_ens = spearmanr(y, y_pred_ensemble)
print(f"SVR+GB平均 (4特征): r_s = {rho_ens:.3f}, p = {p_ens:.4e}")
```
